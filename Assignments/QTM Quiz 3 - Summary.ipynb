```python
result = (
    df[df['colA'] > 10]
    .groupby('group_col', as_index=False).agg
        (average_age = ('value_col','mean'), 
         other_sum = ('other_col','sum'),
        ).query('value_col >= 50')
)
```

3. You will recode a numeric variable into a categorical variable (which takes string values). The category will be based on which interval the numeric variable belongs to.


```python
bins = [0, 50, 70, 100]
labels = ['low', 'medium', 'high']
df['score_category'] = pd.cut(df['score'], bins=bins, labels=labels)
```
