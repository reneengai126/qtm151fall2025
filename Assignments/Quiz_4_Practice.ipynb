1. You will be given a code cell that calculates the coefficients of a quadratic model () between two variables. All code in this cell will be correct. Then in the next code cell, a figure is generated with a scatter plot of the individual data and a line plot of the model prediction. However, the plot is obviously wrong. Make necessary modifications to generate a correct plot.
 

```python
import numpy as np

# Example data
x = np.linspace(0, 10, 20)
y = 3*x**2 + 2*x + 5 + np.random.randn(20)*10

# Fit quadratic model
coeffs = np.polyfit(x, y, 2)
model = np.poly1d(coeffs)
```

```python
import matplotlib.pyplot as plt

# Create fine grid for smooth line
x_fit = np.linspace(x.min(), x.max(), 200)
y_fit = model(x_fit)

plt.scatter(x, y, label='Data')
plt.plot(x_fit, y_fit, color='red', label='Quadratic Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title('Quadratic Model Fit')
plt.show()
```


3. Read in a .csv file to create a DataFrame which contains conversion rates for Euros to U.S. dollars and British Pounds to U.S. dollars over time. Compute non-dimensional percentage growth values for each and plot together on a single plot. (You can use matplotlib.pyplot or the .plot() method connected to a DataFrame).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('conversion_rates.csv', parse_dates=['date'])

# Compute percentage growth relative to the first value
df['EUR_to_USD_growth'] = ((df['EUR_to_USD']-df['EUR_to_USD'].iloc[0])/ df['EUR_to_USD'].iloc[0]) * 100
df['GBP_to_USD_growth'] = ((df['GBP_to_USD']-df['GBP_to_USD'].iloc[0])/ df['GBP_to_USD'].iloc[0]) * 100

# Plot both
df.plot(x='date', y=['EUR_to_USD_growth', 'GBP_to_USD_growth'], figsize=(10,5))
plt.title('Percentage Growth of Conversion Rates')
plt.ylabel('Growth (%)')
plt.show()
```


4. Continuing with the DataFrame of conversion rates, create a new column of Boolean type, which checks whether the conversion rate for one of the currencies was greater than a given value.


 

```python
threshold = 1.20  # example threshold
df['EUR_above_threshold'] = df['EUR_to_USD'] > threshold
```

5. Continuing Problem 4, make a line plot of the conversion rate for this currency and highlight regions where the conversion rate is larger than a given value, using the function "plt.fill_between"

 

```python
plt.figure(figsize=(10,5))
plt.plot(df['date'], df['EUR_to_USD'], label='EUR to USD', color='blue')

plt.fill_between(df['date'],
                 df['EUR_to_USD'],
                 threshold,
                 where=df['EUR_above_threshold'],
                 color='red',
                 alpha=0.3,
                 label=f'Above {threshold}')

plt.axhline(threshold, color='gray', linestyle='--')
plt.legend()
plt.title('EUR to USD with Highlighted High Regions')
plt.show()
```

```python

time = np.linspace(0, 2*np.pi, 200)
df = pd.DataFrame({'time': time})
df['sin'] = np.sin(df['time'])
df['cos'] = np.cos(df['time'])

# Compute numerical derivative
df['d_sin_dt'] = df['sin'].diff() / df['time'].diff()

# Plot comparison
plt.figure(figsize=(8,5))
plt.plot(df['time'], df['cos'], label='cos(t)')
plt.plot(df['time'], df['d_sin_dt'], '--', label='d(sin)/dt (numerical)')
plt.legend()
plt.title('Comparison of cos(t) and Numerical Derivative of sin(t)')
plt.xlabel('time')
plt.show()
```
