4. Write a function that takes in a single number as an input argument, and returns a single string as the output. The value of the output string is one of two choices, which is determined by whether the input satisfies a logical condition. Then use the .apply() method from Pandas to evaluate this function for every row in a DataFrame. You will be told which column/variable should be input to the function.


```python
def age_group(age):
    if age < 18:
        return "Minor"
    else:
        return "Adult"

# Apply function
df["age_category"] = df["age"].apply(age_group)

# Show result
print(df[["age", "age_category"]].head())
```

5. Given a DataFrame, you will want to randomly assign each row to one of two groups. To do so, you will add a new column to the DataFrame called "random_var" which is filled with random numbers from a given distribution. Then you will create two new DataFrames, based on the whether each row's random variable satisfies a given logical condition. These two DataFrames represent the random assignment into two groups.


```python
df["random_var"] = np.random.rand(len(df))

# Assign groups based on condition
group1 = df[df["random_var"] < 0.5]
group2 = df[df["random_var"] >= 0.5]

print(f"Group 1 size: {len(group1)}")
print(f"Group 2 size: {len(group2)}")
```

6. Continuing question 5, you will make two scatter plots from the data in each randomly assigned group. So that they appear side-by-side, you will need to use the subplots command from the matplotlib.pyplot library.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Scatter plot for Group 1
ax1.scatter(group1["age"], group1["income"], color='blue')
ax1.set_title("Group 1: age vs income")
ax1.set_xlabel("Age")
ax1.set_ylabel("Income")

# Scatter plot for Group 2
ax2.scatter(group2["age"], group2["income"], color='green')
ax2.set_title("Group 2: age vs income")
ax2.set_xlabel("Age")
ax2.set_ylabel("Income")

plt.tight_layout()
plt.show()
```
