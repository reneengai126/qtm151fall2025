#### 1. Use an if/elif/else block to test whether a string belongs to one of two lists. Depending on the outcome (belongs to list A, belongs to list B, doesn't belong to either list) print an appropriate message to the screen.


```python
list_A = ["apple", "banana", "cherry"]
list_B = ["dog", "cat", "mouse"]
test_string = "banana"
if (test_string in list_A):
    print("The string belongs to list A.")
elif (test_string in list_B):
    print("The string belongs to list B.")  
else:
    print("The string doesn't belong to either list.")
```

#### 2. You are given two lists of numbers, x_vals and y_vals, and must create a Numpy array where the components are related to those in the lists (for example, you might be asked to create a Numpy array with entries  ). Then generate a histogram of the values in the Numpy array, with labeled axes and a title. You will also be asked to change the color of the bars in the histogram.


```python
x_vals = [1, 2, 3, 4, 5]
y_vals = [10, 20, 30, 40, 50]
import numpy as np
import matplotlib.pyplot as plt
array = np.array(x_vals) * np.array(y_vals)
plt.hist(array, color='blue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of Numpy Array Values')
```

#### 3. Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. Each sub-figure will have its own labeled axes and title.


```python
fig, subfig_array = plt.subplots(nrows = 1, ncols = 2,figsize = (6,3))

normal_samples = np.random.normal(loc=0, scale=1, size=1000)
unif_samples = np.random.uniform(low=0, high=1, size=1000)

subfig_array[0].hist(x = normal_samples)
subfig_array[0].set_title("Normal Distribution\n with mean -2, s.d. 6")
subfig_array[0].set_xlabel("Sample Value")
subfig_array[0].set_ylabel("Frequency")

# Second Figure
subfig_array[1].hist(x = unif_samples)
subfig_array[1].set_title("Uniform Distribution\n over [-3, 5]")
subfig_array[1].set_xlabel("Sample Value")
subfig_array[1].set_ylabel("Frequency")
plt.show()

# Second way to do it 
sample_1 = np.random.normal(loc=0, scale=1, size=1000)
sample_2 = np.random.uniform(low=0, high=1, size=1000)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.hist(sample_1, bins=30, color='blue', edgecolor='black')
ax1.set_title('Histogram of Normal Distribution')
ax1.set_xlabel('Value')
ax1.set_ylabel('Frequency')
ax2.hist(sample_2, bins=30, color='green', edgecolor='black')
ax2.set_title('Histogram of Uniform Distribution')
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')
```

#### 4. Read in a .csv file to create a Pandas DataFrame. Create a scatter plot with one of the variables (or features) corresponding to the x-axis and another corresponding to the y-axis. The plot should have a title and labeled axes.

```python
import pandas as pd
us_emissions = pd.read_csv('united_states_co2_emissions.csv')
us_emissions.head()
plt.scatter(x=us_emissions["year"], y=us_emissions["total_emissions"])
plt.xlabel('Year')
plt.ylabel('Total Emissions (Million Metric Tons)')
plt.title('US CO2 Emissions Over Time')
```

#### 5. Read in a .csv file to create a Pandas DataFrame. Loop over a list of variable/column names, printing the mean of each variable.


```python
for column in ['total_emissions']:
    mean_value = us_emissions[column].mean()
    print(f"The mean of {column} is {mean_value}")

for column in ['year']:
    mean_value = us_emissions[column].mean()
    print(f"The mean of {column} is {mean_value}")
```

#### 6. Create a list of .csv file names and an additional empty list. Then in a for loop, read in each file to make a DataFrame, compute the mean of one of the variables/features/columns, and store it as an element of the originally empty list. Then print the list. 

```python
file_names = ['united_states_co2_emissions.csv', 'china_co2_emissions.csv'] 
mean_values = []
for file in file_names:
    df = pd.read_csv(file)
    mean_value = df['total_emissions'].mean()  
    mean_values.append(float(mean_value))
print("List of Means:", mean_values)
```
